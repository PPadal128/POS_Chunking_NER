```python
a = []
for k in junk:
  p = word_tokenize(k)
  m = pos_tag(p)
  a.append(m)
print(a)
```

**Chuncking_Process After POS tagging**

```python
import nltk
from nltk import RegexpParser


grammer = r"""
    NP: {<DT|PRP\$>?<JJ.*>*<NN.*>+}
    PP: {<IN><NP>}
    VP: {<VB.*><NP|PP|ADJP|ADVP>*}
    ADJP: {<JJ.*>}
    ADVP: {<RB.*>}
"""
```

```python
chunkParse = RegexpParser(grammer)

results = []

for sentence in a:
    result = chunkParse.parse(sentence)
    results.append(result)


for tree in results:
    tree.pretty_print()
    print()
```

**NER = Named Entity Recognition**

```python
from nltk import ne_chunk

nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

for sentence in a:
    result = ne_chunk(sentence)
    result.pretty_print()
    print()
```

```python

import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')


junk = [
    "Barack Obama visited India.",
    "Elon Musk founded SpaceX.",
    "Google opened an office in Bangalore.",
    "Taim Patel lives in Odisha."
]


for sentence in junk:

    print("=" * 60)
    print("Sentence:", sentence)
    print()

    # 1. Tokenization
    tokens = word_tokenize(sentence)

    # 2. POS Tagging
    pos_tags = pos_tag(tokens)

    print("POS Tags:")
    print(pos_tags)
    print()

    # 3. Named Entity Recognition
    ner_tree = ne_chunk(pos_tags)

    print("NER Tree:")
    ner_tree.pretty_print()

    print()
```
